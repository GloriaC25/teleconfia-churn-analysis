```python
# Inspect tables

import sqlalchemy as sa

# Crear engine puente Python-SQL
engine = sa.create_engine("sqlite:///data/telco_churn.db")

# Crear inspector para ver tablas
inspector = sa.inspect(engine)

# Ver nombres de tablas
inspector.get_table_names()


# Read tables using LIMIT and print them to get a short overview
import pandas as pd

print(inspector.get_table_names())

# 4. Crear conexión
connection = engine.connect()

# 5. Leer algunas filas de cada tabla
df_churn_preview = pd.read_sql("SELECT * FROM churn_data LIMIT 5;", connection)
df_cities_preview = pd.read_sql("SELECT * FROM cities LIMIT 5;", connection)

display(df_churn_preview)
display(df_cities_preview)

# Write SQL query to join the tables
sql_query = """
SELECT *
FROM churn_data AS c
LEFT JOIN cities AS ci
ON c.local_area_code = ci.area_code;
"""
# Read complete data using the sql query.
df = pd.read_sql(sql_query, connection)
df.head()

# Close connection
# Cerrar conexión
connection.close()

2) Überprüfung und Reinigung von Daten

df.dtypes

df.isna().sum()

# Muestra únicamente las filas donde TODAS las columnas son NaN (Eliminar?)
df[df.isna().all(axis=1)]

# E1. Eliminar 16 filas donde TODAS las columnas son NaN
df = df.dropna(how="all")

#Comprobación E1-Filas donde TODAS las columnas son NaN
df.isna().sum()

# Drop duplicated columns Eliminar!
# Exploración informacion de columnas
display(df.columns)
display(df.head())
display(df.tail())

# E2.Eliminar la columna 'area_code' porque es redundante.
# Según el Datenwörterbuch, la columna original es 'local_area_code'.
# 'area_code' proviene de la tabla cities y contiene exactamente la misma información.
df = df.drop(columns=["area_code"])

#Comprobación E2-columna 'area_code' porque es redundante
df.columns

# Check the missing values
df.isna().sum()

# Check for wrong values

df.describe(include="all")

# Exploracion customer_service_calls 37 negativos
# Cuenta cuántos valores negativos existen en la columna
(df["customer_service_calls"] < 0).sum()

# Muestra los valores negativos únicos
df.loc[df["customer_service_calls"] < 0, "customer_service_calls"].unique()

# R1- Convertimos TODOS los 37 valores negativos (-1 y -2) en NaN
# porque según el Datenwörterbuch, la cantidad de llamadas al servicio
# NUNCA puede ser negativa.
import numpy as np

df.loc[df["customer_service_calls"] < 0, "customer_service_calls"] = np.nan


# Comprobacion R1 37 valores negativos (-1 y -2) en NaN
df.isna().sum()

# Exploración number_vmail_messages 22 NaN
# Contamos cuántos NaN en number_vmail_messages
# pertenecen a clientes que NO tienen voice_mail_plan = "no"
df[(df["voice_mail_plan"] == "no") & (df["number_vmail_messages"].isna())].shape[0]

# Mostramos únicamente las columnas voice_mail_plan y number_vmail_messages
# para las 22 filas donde number_vmail_messages es NaN
# y el cliente NO tiene plan de buzón de voz.
df.loc[
    (df["voice_mail_plan"] == "no") & (df["number_vmail_messages"].isna()),
    ["voice_mail_plan", "number_vmail_messages"]
]

#R2- Reemplazamos los 22 NaN en number_vmail_messages con 0
# EXCLUSIVAMENTE para los clientes que NO tienen plan de buzón de voz (voice_mail_plan = "no")
# porque según el Datenwörterbuch, un cliente sin plan NO puede tener mensajes de voz.
df.loc[
    (df["voice_mail_plan"] == "no") & (df["number_vmail_messages"].isna()),
    "number_vmail_messages"
] = 0

# Comprobacion R2 22 NaN en number_vmail_messages con 0
df.isna().sum()

# Exploracion 11 NaN total_days_calls
# existen minutos y cargos (lo que indicaría que sí hubo llamadas)
df.loc[
    df["total_day_calls"].isna(),
    ["total_day_calls", "total_day_minutes", "total_day_charge"]
]


# Exploración 8 NaN en total_eve_calls
# existen minutos y cargos (lo que indicaría que sí hubo llamadas)
df.loc[
    df["total_eve_calls"].isna(),
    ["total_eve_calls", "total_eve_minutes", "total_eve_charge"]
]

# Exploración 5 NaN total_night_calls
# existen minutos y cargos (lo que indicaría que sí hubo llamadas)
df.loc[
    df["total_night_calls"].isna(),
    ["total_night_calls", "total_night_minutes", "total_night_charge"]
]

# Creamos columnas indicadoras SOLO para las 4 columnas que aún no tienen flag
df["missing_total_day_calls"] = df["total_day_calls"].isna().astype(int)
df["missing_total_eve_calls"] = df["total_eve_calls"].isna().astype(int)
df["missing_total_night_calls"] = df["total_night_calls"].isna().astype(int)
df["missing_customer_service_calls"] = df["customer_service_calls"].isna().astype(int)

# I1- Imputamos con la mediana de cada columna
df["total_day_calls"].fillna(df["total_day_calls"].median(), inplace=True)
df["total_eve_calls"].fillna(df["total_eve_calls"].median(), inplace=True)
df["total_night_calls"].fillna(df["total_night_calls"].median(), inplace=True)
df["customer_service_calls"].fillna(df["customer_service_calls"].median(), inplace=True)

# Comprobación I1-Imputamos con la mediana
df[[
    "total_day_calls",
    "total_eve_calls",
    "total_night_calls",
    "customer_service_calls"
]].isna().sum()

df.dtypes

# Conversión de tipos según Datenwörterbuch

# Numéricos enteros
df["account_length"] = df["account_length"].astype(int)
df["number_vmail_messages"] = df["number_vmail_messages"].astype(int)
df["total_day_calls"] = df["total_day_calls"].astype(int)
df["total_eve_calls"] = df["total_eve_calls"].astype(int)
df["total_night_calls"] = df["total_night_calls"].astype(int)
df["customer_service_calls"] = df["customer_service_calls"].astype(int)

# Variables categóricas
df["international_plan"] = df["international_plan"].astype("category")
df["voice_mail_plan"] = df["voice_mail_plan"].astype("category")
df["churn"] = df["churn"].astype(int).astype("category")
df["local_area_code"] = df["local_area_code"].astype(int).astype("category")
df["phone_num"] = df["phone_num"].astype(str)
df["city"] = df["city"].astype("category")

df.info()

3) Wie lauten die Namen der vier Stadtgebiete mit der höchsten Kundenabwanderung?

# Calculate overall churn ratio in the dataset

# 1. Mostrar dtype original
print("Tipo original de churn:", df['churn'].dtype)

# 2. Convertir churn a entero
df['churn'] = df['churn'].astype(int)

# 3. Calcular churn promedio (proporción)
overall_churn_rate = df['churn'].mean()

# 4. Mostrar resultados
print("Churn promedio (proporción):", overall_churn_rate)
print("Churn promedio (%):", round(overall_churn_rate * 100, 2))

# Calculate churn ratio based on the city
# ---------------------------------------------------------
# Calcular la tasa de churn por ciudad
# ---------------------------------------------------------

# 1. Crear una tabla cruzada con valores absolutos
#    Esto muestra cuántos clientes churn=0 y churn=1 hay por ciudad
ct_abs = pd.crosstab(df['city'], df['churn'])

# 2. Crear una tabla cruzada con proporciones (tasa de churn por ciudad)
#    normalize='index' → calcula proporciones por fila (por ciudad)
ct_rate = pd.crosstab(df['city'], df['churn'], normalize='index')

# 3. Extraer solo la columna churn=1 (clientes que se fueron)
#    y ordenar de mayor a menor
city_churn_rate = ct_rate[1].sort_values(ascending=False)

# 4. Mostrar las 4 ciudades con mayor churn
top4_cities = city_churn_rate.head(4)
top4_cities


# Verificar si las ciudades están bien representadas
# ---------------------------------------------------------
# Contar cuántos clientes hay por ciudad
city_counts = df['city'].value_counts()

city_counts

4a) Welche kategorische Variable eignet sich als Indikator für Kundenabwanderung? Identifiziere alle Kund:innen, die auf dieser Basis kontaktiert werden sollten

# Check 'international_plan'

# 1. Crosstab con valores absolutos
ct_int_abs = pd.crosstab(df['international_plan'], df['churn'])
print("Crosstab valores absolutos:")
display(ct_int_abs)

# 2. Crosstab con proporciones
ct_int_rate = pd.crosstab(df['international_plan'], df['churn'], normalize='index')
print("\nCrosstab proporciones:")
display(ct_int_rate)

# Check 'voice_mail_plan'

# 1. Crosstab con valores absolutos
ct_vm_abs = pd.crosstab(df['voice_mail_plan'], df['churn'])
print("Crosstab valores absolutos:")
display(ct_vm_abs)

# 2. Crosstab con proporciones (muy importante para Tipp #2)
ct_vm_rate = pd.crosstab(df['voice_mail_plan'], df['churn'], normalize='index')
print("\nCrosstab proporciones:")
display(ct_vm_rate)

# ---------------------------------------------------------
# Tabla comparativa de churn por variable categórica
# con porcentajes, número de clientes y gráfica comparativa
# ---------------------------------------------------------

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Churn promedio del dataset
churn_mean = df['churn'].mean() * 100

# 2. Crosstab con porcentajes
ct_int_pct = pd.crosstab(df['international_plan'], df['churn'], normalize='index') * 100
ct_vm_pct  = pd.crosstab(df['voice_mail_plan'], df['churn'], normalize='index') * 100

# 3. Crosstab con conteos absolutos
ct_int_abs = pd.crosstab(df['international_plan'], df['churn'])
ct_vm_abs  = pd.crosstab(df['voice_mail_plan'], df['churn'])

# 4. Crear tabla resumen
tabla_comparativa = pd.DataFrame({
    'Churn promedio (%)': [churn_mean, churn_mean],

    'Churn con categoría (%)': [
        ct_int_pct.loc['yes', 1],
        ct_vm_pct.loc['yes', 1]
    ],
    'Churn sin categoría (%)': [
        ct_int_pct.loc['no', 1],
        ct_vm_pct.loc['no', 1]
    ],

    'Clientes con categoría (yes)': [
        ct_int_abs.loc['yes'].sum(),
        ct_vm_abs.loc['yes'].sum()
    ],
    'Clientes sin categoría (no)': [
        ct_int_abs.loc['no'].sum(),
        ct_vm_abs.loc['no'].sum()
    ]
}, index=['international_plan', 'voice_mail_plan'])

# Mostrar tabla
display(tabla_comparativa.round(2))

# ---------------------------------------------------------
# Gráfica comparativa
# ---------------------------------------------------------

plt.figure(figsize=(10,6))

# Preparar datos para la gráfica
plot_df = tabla_comparativa[['Churn con categoría (%)', 'Churn sin categoría (%)']].copy()
plot_df = plot_df.rename(columns={
    'Churn con categoría (%)': 'Con categoría (yes)',
    'Churn sin categoría (%)': 'Sin categoría (no)'
})

plot_df.plot(kind='bar', figsize=(10,6), color=['steelblue', 'salmon'])

plt.title('Comparación de churn por variable categórica')
plt.ylabel('Churn (%)')
plt.xlabel('Variable categórica')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend(title='Grupo')

plt.show()

# Select customers based on categorical data

# 1. Filtrar clientes con international_plan = 'yes'
mask_plan = df['international_plan'] == 'yes'

# 2. Filtrar clientes que NO han hecho churn (churn = 0)
mask_no_churn = df['churn'] == 0

# 3. Combinar ambas condiciones con AND lógico
at_risk_cat = df[mask_plan & mask_no_churn]

# 4. Mostrar cuántos clientes cumplen la condición
print("Número de clientes en riesgo según international_plan:", len(at_risk_cat))

# 5. Mostrar la tabla resultante
at_risk_cat

4b) Welche ganzzahlige Variable eignet sich als Indikator für Kundenabwanderung? Wie würdest du den Grenzwert setzen? Identifiziere die Kund:innen, die auf dieser Basis kontaktiert werden sollten.

# How many unique values are in each column?

# Lista de columnas que parecen enteras
int_columns = [
    'account_length',
    'number_vmail_messages',
    'total_day_calls',
    'total_eve_calls',
    'total_night_calls',
    'customer_service_calls'
]

# Recorrer cada columna y mostrar cuántos valores únicos tiene
for col in int_columns:
    # nunique() cuenta cuántos valores distintos hay en la columna
    print(col, "→", df[col].nunique(), "valores únicos")

# ---------------------------------------------------------
# Tabla resumen de columnas enteras para Tarea 4b
# ---------------------------------------------------------

import pandas as pd

# Columnas enteras identificadas por el Notebook
int_columns = [
    'account_length',
    'number_vmail_messages',
    'total_day_calls',
    'total_eve_calls',
    'total_night_calls',
    'customer_service_calls'
]

# Crear listas para la tabla
unique_counts = []
variable_type = []
analysis_type = []
justification = []

for col in int_columns:
    n_unique = df[col].nunique()
    unique_counts.append(n_unique)

    # Clasificación según cantidad de valores únicos
    if n_unique < 20:
        variable_type.append("Categórica")
        analysis_type.append("Crosstab + Barplot")
        justification.append("Pocos valores únicos → se comporta como categórica")
    else:
        variable_type.append("Continua")
        analysis_type.append("Boxplot / Histograma")
        justification.append("Muchos valores únicos → distribución continua")

# Crear DataFrame resumen
tabla_resumen = pd.DataFrame({
    'Valores únicos': unique_counts,
    'Tipo de variable': variable_type,
    'Tipo de análisis recomendado': analysis_type,
    'Justificación': justification
}, index=int_columns)

tabla_resumen

# Check column that can be treated as a categorical one

#Analizar la columna categórica customer_service_calls
# ---------------------------------------------------------
import matplotlib.pyplot as plt

# 1. Tabla cruzada con valores absolutos
ct_abs = pd.crosstab(df['customer_service_calls'], df['churn'])

print("Crosstab – valores absolutos:")
display(ct_abs)

# 2. Tabla cruzada con porcentajes por fila
ct_rel = pd.crosstab(df['customer_service_calls'], df['churn'], normalize='index') * 100

print("\nCrosstab – porcentajes (%):")
display(ct_rel.round(2))

# 3. Visualización: churn (%) por número de llamadas al servicio
plt.figure(figsize=(10, 5))
ct_rel[1].plot(kind='bar', color='steelblue')

plt.title('Churn según número de llamadas al servicio (customer_service_calls)')
plt.xlabel('Número de llamadas al servicio')
plt.ylabel('Churn (%)')
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.show()

# Check other integer columns with boxplots or histograms

#Histogramas para columnas enteras continuas
# ---------------------------------------------------------

import matplotlib.pyplot as plt

columnas_continuas = [
    'account_length',
    'number_vmail_messages',
    'total_day_calls',
    'total_eve_calls',
    'total_night_calls'
]

# Crear un histograma por cada columna, separado por churn
for col in columnas_continuas:
    plt.figure(figsize=(8, 4))

    # Histograma para churn = 0
    plt.hist(df[df['churn'] == 0][col], bins=30, alpha=0.6, label='churn = 0', color='steelblue')

    # Histograma para churn = 1
    plt.hist(df[df['churn'] == 1][col], bins=30, alpha=0.6, label='churn = 1', color='orange')

    plt.title(f'Distribución de {col} según churn')
    plt.xlabel(col)
    plt.ylabel('Frecuencia')
    plt.legend()
    plt.grid(axis='y', linestyle='--', alpha=0.4)

    plt.show()

# Select customers based on integer column

#Seleccionar clientes en riesgo según customer_service_calls
# ---------------------------------------------------------

# Definir el umbral
umbral = 4

# Filtrar clientes que:
# 1) No han hecho churn (churn = 0)
# 2) Tienen customer_service_calls >= 4
clientes_en_riesgo = df[(df['churn'] == 0) & (df['customer_service_calls'] >= umbral)]

# Mostrar cuántos clientes cumplen la condición
print("Número de clientes en riesgo:", len(clientes_en_riesgo))

# Mostrar las primeras filas
clientes_en_riesgo.head()

4c) Welche Fließkomma-Variable könnte sich als Indikator für Kundenabwanderung eignen? Bestimme den Grenzwert hier mit einer logistischen Regression. Identifiziere die Kund:innen, die auf dieser Basis kontaktiert werden sollten.

# Check ..._charge and ..._minutes columns;  Is there any kind of linear depency?

# Objetivo:
# Comprobar si existe dependencia lineal entre las columnas
# ..._minutes y ..._charge para day, eve y night.
#
# Si los puntos forman una línea recta → contienen la misma
# información → solo necesitamos UNA de ellas para la regresión.
# ------------------------------------------------------------

import matplotlib.pyplot as plt

# Crear figura con 3 subplots (day, eve, night)
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# ------------------------------------------------------------
# Scatterplot 1: total_day_minutes vs total_day_charge
# ------------------------------------------------------------
axes[0].scatter(df['total_day_minutes'], df['total_day_charge'], alpha=0.5)
axes[0].set_title('Day: minutes vs charge')
axes[0].set_xlabel('total_day_minutes')
axes[0].set_ylabel('total_day_charge')
# Si los puntos forman una línea → dependencia lineal fuerte

# ------------------------------------------------------------
# Scatterplot 2: total_eve_minutes vs total_eve_charge
# ------------------------------------------------------------
axes[1].scatter(df['total_eve_minutes'], df['total_eve_charge'], alpha=0.5)
axes[1].set_title('Evening: minutes vs charge')
axes[1].set_xlabel('total_eve_minutes')
axes[1].set_ylabel('total_eve_charge')
# Si los puntos forman una línea → contienen la misma información

# ------------------------------------------------------------
# Scatterplot 3: total_night_minutes vs total_night_charge
# ------------------------------------------------------------
axes[2].scatter(df['total_night_minutes'], df['total_night_charge'], alpha=0.5)
axes[2].set_title('Night: minutes vs charge')
axes[2].set_xlabel('total_night_minutes')
axes[2].set_ylabel('total_night_charge')
# Si los puntos forman una línea → redundancia entre minutes y charge

plt.tight_layout()
plt.show()

import seaborn as sns

sns.scatterplot(data=df, x='total_day_minutes', y='total_day_charge')

# Identifying outliers and deal with them-Valores atipicos
# ------------------------------------------------------------
import pandas as pd

# 1. Top 20 day minutes
top_day = df[['total_day_minutes', 'total_day_charge', 'churn']] \
    .sort_values('total_day_minutes', ascending=False) \
    .head(20)

# 2. Top 20 evening minutes
top_eve = df[['total_eve_minutes', 'total_eve_charge', 'churn']] \
    .sort_values('total_eve_minutes', ascending=False) \
    .head(20)

# 3. Top 20 night minutes
top_night = df[['total_night_minutes', 'total_night_charge', 'churn']] \
    .sort_values('total_night_minutes', ascending=False) \
    .head(20)

# Renombrar columnas para evitar confusión visual
top_day.columns = ['day_minutes', 'day_charge', 'churn']
top_eve.columns = ['eve_minutes', 'eve_charge', 'churn']
top_night.columns = ['night_minutes', 'night_charge', 'churn']

# 4. Combinar las 3 tablas lado a lado
combined_tables = pd.concat([top_day.reset_index(drop=True),
                             top_eve.reset_index(drop=True),
                             top_night.reset_index(drop=True)], axis=1)

# 5. Mostrar tabla final
combined_tables

# Boxplots or histograms relevant columns
# Boxplots para comparar churn=0 vs churn=1
# en total_day_minutes, total_eve_minutes y total_night_minutes.
# ------------------------------------------------------------

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(18, 6))

# ------------------------------------------------------------
# Boxplot 1: total_day_minutes vs churn
# ------------------------------------------------------------
plt.subplot(1, 3, 1)
sns.boxplot(data=df, x='churn', y='total_day_minutes')
plt.title('Day Minutes vs Churn')
plt.xlabel('Churn (0 = stay, 1 = leave)')
plt.ylabel('total_day_minutes')

# ------------------------------------------------------------
# Boxplot 2: total_eve_minutes vs churn
# ------------------------------------------------------------
plt.subplot(1, 3, 2)
sns.boxplot(data=df, x='churn', y='total_eve_minutes')
plt.title('Evening Minutes vs Churn')
plt.xlabel('Churn (0 = stay, 1 = leave)')
plt.ylabel('total_eve_minutes')

# ------------------------------------------------------------
# Boxplot 3: total_night_minutes vs churn
# ------------------------------------------------------------
plt.subplot(1, 3, 3)
sns.boxplot(data=df, x='churn', y='total_night_minutes')
plt.title('Night Minutes vs Churn')
plt.xlabel('Churn (0 = stay, 1 = leave)')
plt.ylabel('total_night_minutes')

plt.tight_layout()
plt.show()

# Pair plot

# Pair plot para visualizar correlaciones y distribuciones
# ------------------------------------------------------------

import seaborn as sns
import matplotlib.pyplot as plt

# Seleccionamos solo las columnas relevantes
cols = ['total_day_minutes', 'total_eve_minutes', 'total_night_minutes', 'churn']

sns.pairplot(df[cols], hue='churn', palette={0: 'blue', 1: 'orange'})
plt.show()

# Correlation matrix

# Correlation matrix + heatmap
# ------------------------------------------------------------

import seaborn as sns
import matplotlib.pyplot as plt

# Seleccionamos solo las columnas numéricas relevantes
cols = ['total_day_minutes', 'total_eve_minutes', 'total_night_minutes', 'churn']

# Calculamos la matriz de correlación
corr_matrix = df[cols].corr()

# Creamos una figura grande para evitar solapamientos
plt.figure(figsize=(10, 8))

# Dibujamos la heatmap
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)

plt.title('Correlation Matrix — Minutes vs Churn')
plt.show()

# Drop redundant column(s)

# Drop redundant columns
# ------------------------------------------------------------

df = df.drop(columns=[
    'total_day_charge',
    'total_eve_charge',
    'total_night_charge'
])

# Logistic regression

# Define a logit model and fit it

# Check model summary

# ------------------------------------------------------------
# LOGISTIC REGRESSION
# ------------------------------------------------------------

import statsmodels.formula.api as smf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Asegurar que churn es numérico
df['churn'] = df['churn'].astype(int)

# 2. Definir y ajustar el modelo logístico
result = smf.logit('churn ~ total_day_minutes', data=df).fit()

# 3. Mostrar resumen
print(result.summary())

# 4. Crear rango de valores y predecir probabilidades
x_range = np.linspace(df['total_day_minutes'].min(),
                      df['total_day_minutes'].max(),
                      300)
predict_df = pd.DataFrame({'total_day_minutes': x_range})
y_pred = result.predict(predict_df)

# 5. Graficar S-curve
plt.figure(figsize=(10,6))
plt.plot(x_range, y_pred, color='blue')
plt.axhline(0.5, color='red', linestyle='--')
plt.xlabel('total_day_minutes')
plt.ylabel('Probabilidad de churn')
plt.title('S-curve — Logistic Regression')
plt.grid(True)
plt.show()

# 6. Calcular threshold
threshold = -result.params['Intercept'] / result.params['total_day_minutes']
print("Threshold (p=0.5):", threshold)

# Use the trained results to predict a given range and plot the S-curve

# ------------------------------------------------------------

# 1. Crear un rango de valores entre 0 y 800
x_range = np.linspace(0, 800, 300)

# 2. Crear un DataFrame con el nombre correcto de la columna
predict_df = pd.DataFrame({'total_day_minutes': x_range})

# 3. Predecir probabilidades usando el modelo entrenado
y_pred = result.predict(predict_df)

# 4. Graficar la S-curve
plt.figure(figsize=(10,6))
plt.plot(x_range, y_pred, color='blue', label='Predicted probability')

# 5. Línea horizontal en 50%
plt.axhline(0.5, color='red', linestyle='--', label='50% threshold')

# 6. Etiquetas y título
plt.xlabel('total_day_minutes')
plt.ylabel('Probabilidad de churn')
plt.title('S-curve — Logistic Regression')
plt.legend()
plt.grid(True)
plt.show()

# Calculate treshold value and plot it as a vertical line in logistic regression plot

# 1. Calcular el threshold (ya lo tienes, pero lo incluimos por claridad)
coef_intercept = result.params['Intercept']
coef_minutes = result.params['total_day_minutes']
threshold = -coef_intercept / coef_minutes

# 2. Crear un rango extendido para la curva
x_range = np.linspace(0, 800, 300)
predict_df = pd.DataFrame({'total_day_minutes': x_range})
y_pred = result.predict(predict_df)

# 3. Graficar la S-curve
plt.figure(figsize=(10,6))
plt.plot(x_range, y_pred, color='blue', label='Predicted probability')

# 4. Línea horizontal del 50%
plt.axhline(0.5, color='red', linestyle='--', label='50% threshold')

# 5. Línea vertical del threshold calculado
plt.axvline(threshold, color='green', linestyle='--', label=f'Threshold = {threshold:.2f}')

# 6. Etiquetas y título
plt.xlabel('total_day_minutes')
plt.ylabel('Probabilidad de churn')
plt.title('S-curve with Threshold — Logistic Regression')
plt.legend()
plt.grid(True)
plt.show()

# Identify customer(s) to be contacted based on critical value from logistic regression
# 1. Crear una columna con la predicción basada en el threshold
df['predicted_churn'] = (df['total_day_minutes'] > threshold).astype(int)

# 2. Filtrar clientes que NO han hecho churn (churn == 0)
#    pero que el modelo predice que SÍ harán churn (predicted_churn == 1)
customers_to_contact = df[(df['churn'] == 0) & (df['predicted_churn'] == 1)]

# 3. Mostrar los clientes en riesgo
customers_to_contact

5) Visualisierung der Stadtgebiete und der drei anderen ausgewählten Datenreihen

# Column chart of urban areas


import matplotlib.pyplot as plt

# 1. Calcular churn rate por ciudad (en %)
city_churn = df.groupby('city')['churn'].mean().sort_values(ascending=False) * 100

# 2. Calcular número de clientes por ciudad
city_counts = df['city'].value_counts().loc[city_churn.index]

# 3. Calcular churn promedio (en %)
avg_churn = df['churn'].mean() * 100

# 4. Identificar el Top 4
top4_cities = city_churn.head(4).index

# 5. Asignar colores: rojo para Top 4, azul para el resto
colors = ['#D00000' if city in top4_cities else '#2A6F97' for city in city_churn.index]

# 6. Crear figura con fondo blanco
plt.figure(figsize=(14, 7), facecolor='white')

# 7. Gráfico de barras
bars = plt.bar(
    city_churn.index,
    city_churn.values,
    color=colors,
    width=0.6
)

# 8. Línea horizontal del churn promedio
plt.axhline(
    avg_churn,
    color='#495057',
    linestyle='--',
    linewidth=2,
    label=f'Durchschnitt: {avg_churn:.1f}%'
)

# 9. Etiquetas encima de cada barra (churn + número de clientes)
for i, (value, count) in enumerate(zip(city_churn, city_counts)):
    plt.text(
        i,
        value + (city_churn.max() * 0.03),
        f'{value:.1f}%\n({count} Kunden)',
        ha='center',
        fontsize=10,
        color='#343A40'
    )

# 10. Ajustar límites del eje Y
plt.ylim(0, city_churn.max() * 1.30)

# 11. Títulos y etiquetas
plt.title(
    'Abwanderungsrate nach Stadtgebiet (%) — mit Kundenanzahl',
    fontsize=18,
    fontweight='bold',
    color='#343A40',
    pad=20
)
plt.xlabel('Stadtgebiet', fontsize=13, color='#343A40')
plt.ylabel('Abwanderungsrate (%)', fontsize=13, color='#343A40')

# 12. Rotar etiquetas del eje X
plt.xticks(rotation=30, ha='right', fontsize=11, color='#343A40')

# 13. Leyenda
plt.legend(fontsize=12)

# 14. Cuadrícula suave
plt.grid(axis='y', linestyle='--', alpha=0.25)

plt.tight_layout()
plt.show()

# Column chart of categorical data

import matplotlib.pyplot as plt
from IPython.display import HTML, display

# 1. Cálculos
cat_churn = df.groupby('international_plan')['churn'].mean().sort_values(ascending=False) * 100
churn_counts = df.groupby('international_plan')['churn'].sum()
avg_churn = df['churn'].mean() * 100

# Clientes en riesgo: tienen International Plan = yes y churn = 0
at_risk_cat = df[(df['international_plan'] == 'yes') & (df['churn'] == 0)]
n_risk = len(at_risk_cat)

# ------------------------------------------------------------
# GRAFICA ARRIBA
# ------------------------------------------------------------
plt.figure(figsize=(12, 6), facecolor='white')

bar_color = '#2A6F97'
avg_line_color = '#495057'
text_color = '#343A40'

plt.bar(cat_churn.index.astype(str), cat_churn.values, color=bar_color, width=0.55)

# Línea horizontal del churn promedio
plt.axhline(avg_churn, color=avg_line_color, linestyle='--', linewidth=2,
            label=f'Durchschnitt: {avg_churn:.1f}%')

# Etiquetas encima de cada barra
for i, (value, n) in enumerate(zip(cat_churn, churn_counts)):
    plt.text(
        i,
        value + (cat_churn.max() * 0.02),
        f'{value:.1f}%  (n={n})',
        ha='center',
        fontsize=11,
        color=text_color
    )

plt.ylim(0, cat_churn.max() * 1.25)

plt.title(
    'Abwanderungsrate nach International Plan (%)',
    fontsize=18, fontweight='bold', color=text_color, pad=20
)
plt.xlabel('International Plan (Yes/No)', fontsize=13, color=text_color)
plt.ylabel('Abwanderungsrate (%)', fontsize=13, color=text_color)

plt.xticks(fontsize=11, color=text_color)
plt.legend(fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.25)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# KPI ABAJO
# ------------------------------------------------------------
display(HTML(f"""
<div style="text-align:center; font-family:Arial; margin-top:25px;">
    <div style="font-size:60px; font-weight:bold; color:#2A6F97;">
        {n_risk}
    </div>
    <div style="font-size:20px; color:#343A40; margin-bottom:10px;">
        Kunden mit hohem Risiko (International Plan = Yes, Churn = 0)
    </div>
</div>
"""))

# ------------------------------------------------------------
# TABLA DE CLIENTES EN RIESGO
# ------------------------------------------------------------
display(at_risk_cat)

# Column chart of integer data


import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import HTML, display

# -----------------------------
# Preparación de datos
# -----------------------------
# Churn rate por número de llamadas al servicio
calls_churn = df.groupby('customer_service_calls')['churn'].mean() * 100
counts = df['customer_service_calls'].value_counts().sort_index()

# -----------------------------
# Gráfica
# -----------------------------
plt.figure(figsize=(12, 6), facecolor='white')

bar_color = '#2A6F97'
threshold_color = '#D62828'
text_color = '#343A40'

plt.bar(calls_churn.index, calls_churn.values, color=bar_color, width=0.6)

# Línea vertical del umbral
plt.axvline(4, color=threshold_color, linestyle='--', linewidth=2,
            label='Grenzwert: 4 Anrufe')

# Etiquetas de porcentaje arriba de cada barra
for x, y in zip(calls_churn.index, calls_churn.values):
    plt.text(x, y + 0.5, f'{y:.1f}%', ha='center', fontsize=11, color=text_color)

plt.title(
    'Abwanderungsrate nach Anzahl der Kundenservice-Anrufe\n(Grenzwert = 4 Anrufe)',
    fontsize=18, fontweight='bold', color=text_color, pad=20
)
plt.xlabel('Anzahl der Kundenservice-Anrufe', fontsize=13, color=text_color)
plt.ylabel('Abwanderungsrate (%)', fontsize=13, color=text_color)

plt.xticks(calls_churn.index, rotation=45, fontsize=11, color=text_color)
plt.yticks(fontsize=11, color=text_color)

plt.grid(axis='y', linestyle='--', alpha=0.25)
plt.legend(fontsize=12)

plt.tight_layout()
plt.show()

# -----------------------------
# KPI ABAJO
# -----------------------------
at_risk_int = df[(df['customer_service_calls'] >= 4) & (df['churn'] == 0)]
n_risk_int = len(at_risk_int)

display(HTML(f"""
<div style="text-align:center; font-family:Arial; margin-top:25px;">
    <div style="font-size:60px; font-weight:bold; color:#2A6F97;">
        {n_risk_int}
    </div>
    <div style="font-size:20px; color:#343A40; margin-bottom:10px;">
        Kunden mit hohem Risiko (≥ 4 Service-Anrufe, Churn = 0)
    </div>
</div>
"""))

# -----------------------------
# TABLA DE CLIENTES EN RIESGO
# -----------------------------
display(at_risk_int)

# Line plot for findings in logistic regression
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import HTML, display

# ------------------------------------------------------------
# 1. Preparar datos para la curva S
# ------------------------------------------------------------
x_range = np.linspace(0, 800, 300)
predict_df = pd.DataFrame({'total_day_minutes': x_range})
y_pred = result.predict(predict_df)

# Threshold real calculado en 4c
threshold = 350.74297079545426

# ------------------------------------------------------------
# 2. Gráfica profesional (estilo consistente con Bloques 1–3)
# ------------------------------------------------------------
plt.figure(figsize=(12, 6), facecolor='white')

line_color = '#2A6F97'       # azul petróleo
threshold_color = '#D62828'  # rojo
text_color = '#343A40'       # gris oscuro

# Curva S
plt.plot(x_range, y_pred, color=line_color, linewidth=3, label='Predicted probability')

# Línea horizontal del 50%
plt.axhline(0.5, color=threshold_color, linestyle='--', linewidth=2, label='50% Schwelle')

# Línea vertical del threshold
plt.axvline(threshold, color='#2A9D8F', linestyle='--', linewidth=2,
            label=f'Grenzwert = {threshold:.2f} Minuten')

# Títulos y ejes
plt.title(
    'Wahrscheinlichkeit der Abwanderung nach Tagesminuten\n(Logistische Regression)',
    fontsize=18, fontweight='bold', color=text_color, pad=20
)
plt.xlabel('total_day_minutes', fontsize=13, color=text_color)
plt.ylabel('Abwanderungswahrscheinlichkeit', fontsize=13, color=text_color)

plt.xticks(fontsize=11, color=text_color)
plt.yticks(fontsize=11, color=text_color)

# Grid suave
plt.grid(axis='y', linestyle='--', alpha=0.25)

# Leyenda
plt.legend(fontsize=12)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 3. KPI — Número de clientes en riesgo
# ------------------------------------------------------------

# Crear columna de predicción basada en threshold
df['predicted_churn'] = (df['total_day_minutes'] > threshold).astype(int)

# Filtrar clientes en riesgo: churn = 0 pero modelo = 1
customers_to_contact = df[(df['churn'] == 0) & (df['predicted_churn'] == 1)]
n_risk = len(customers_to_contact)

# KPI visual
display(HTML(f"""
<div style="text-align:center; font-family:Arial; margin-top:25px;">
    <div style="font-size:60px; font-weight:bold; color:#2A6F97;">
        {n_risk}
    </div>
    <div style="font-size:20px; color:#343A40; margin-bottom:10px;">
        Kunden mit hohem Risiko (Tagesminuten > {threshold:.2f}, Churn = 0)
    </div>
</div>
"""))

# ------------------------------------------------------------
# 4. Tabla de clientes en riesgo
# ------------------------------------------------------------
display(customers_to_contact)

```
